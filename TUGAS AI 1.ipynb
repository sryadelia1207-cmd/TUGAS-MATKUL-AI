## 1. Studi Kasus: Metode Induced Polarization (IP) untuk Eksplorasi Mineral Emas/Tembaga
Bertindaklah sebagai **[Role]** Geofisikawan Eksplorasi Mineral Deposit. **[Task]** Evaluasilah data Chargeability (kemampuan menyimpan muatan) dan Resistivitas 2D untuk memetakan zona mineralisasi sulfida yang berpotensi membawa bijih tembaga atau emas porfir. **[Context]** Tim eksplorasi sedang memetakan zona alterasi hidrotermal bawah permukaan untuk menentukan target pengeboran inti (*coring*). Berdasarkan **[Input]** penampang IP 2D yang menunjukkan zona anomali chargeability tinggi (>30 mV/V) berhimpitan dengan zona resistivitas sedang-tinggi (300-800 $\Omega m$) pada kedalaman 50–200 meter, kerjakan dengan **[Constraints]** mengaitkan sifat fisik chargeability terhadap mekanisme polarisasi terinduksi pada mineral sulfida terdiseminasi (seperti pirit atau kalkopirit). Susun **[Output]** berupa laporan evaluasi prospek mineral yang memuat garis batas zona cebakan dan rekomendasi titik koordinat bor eksplorasi. **[Quality Criteria]** Penjelasan harus secara akurat menguraikan mengapa nilai chargeability tinggi menandakan akumulasi mineral logam terdiseminasi di dalam batuan pembawa. Sebagai langkah konfirmasi, **[Optional Verification]** rekomendasikan pengujian sampel inti bor (*core sample*) menggunakan analisis petrografi dan assay geokimia.

## 2. Studi Kasus: Mikroseismik HVSR untuk Mikrozonasi Kerentanan Gempa Bumi
Kamu adalah **[Role]** Ahli Seismologi Teknik dan Mitigasi Bencana Alam. **[Task]** Analisislah kurva rasio spektral HVSR (*Horizontal-to-Vertical Spectral Ratio*) dari data mikrotremor untuk menghitung frekuensi dominan ($f_0$) dan indeks kerentanan seismik ($K_g$). **[Context]** Pemerintah daerah meminta pemetaan wilayah rawan guncangan gempa pada kawasan permukiman padat yang berdiri di atas endapan sedimen kuarter. Diberikan **[Input]** data mikrotremor 3-komponen yang menghasilkan frekuensi puncak dominan $f_0 = 1.2\text{ Hz}$ dan faktor amplifikasi $A_0 = 4.5$, terapkan **[Constraints]** dengan menggunakan persamaan indeks kerentanan $K_g = \frac{A_0^2}{f_0}$ secara eksplisit serta jelaskan implikasi ketebalan sedimen lunak terhadap nilai $f_0$. Hasilkan **[Output]** berupa ringkasan penilaian risiko guncangan gempa dan pemetaan tingkat kerentanan tapak (*site vulnerability index*). **[Quality Criteria]** Analisis harus secara logis mengaitkan nilai frekuensi rendah ($f_0$ kecil) dengan ketebalan lapisan tanah lunak yang memicu pembesaran gelombang seismik (*amplification*). Terakhir, **[Optional Verification]** sarankan validasi nilai kecepatan gelombang geser $V_{s30}$ menggunakan metode MASW (*Multichannel Analysis of Surface Waves*).

## 3. Studi Kasus: Magnetotellurik (MT) untuk Penentuan Dapur Magma / Panas Bumi Dalam
Bertindaklah sebagai **[Role]** Geofisikawan Komputasi dan Pemodelan Elektromagnetik. **[Task]** Interpretasikan penampang resistivitas 2D/3D hasil inversi Magnetotellurik (MT) untuk mendeteksi kedalaman zona konduktif (*conductor body*) yang mengindikasikan keberadaan sumber panas (*heat source*) atau dapur magma. **[Context]** Penelitian geofisika regional ini dilakukan di kompleks gunung api aktif untuk memetakan arsitektur sistem magmatik dan potensi panas bumi dalam. Berdasarkan **[Input]** penampang inversi MT yang memperlihatkan anomali amat konduktif (< 5 $\Omega m$) pada kedalaman 7 hingga 12 km di bawah puncak gunung api, patuhi **[Constraints]** untuk memfokuskan analisis pada sifat elektromagnetik batuan parsial meleleh (*partial melt*) dan fluida kaya ion pada suhu tinggi. Berikan **[Output]** berupa ulasan ilmiah mengenai estimasi kedalaman, geometri, dan perkiraan volume zona konduktif tersebut. **[Quality Criteria]** Argumentasi harus secara fisik menghubungkan penurunan nilai resistivitas yang ekstrem dengan keberadaan lelehan silikat jenuh fluida pada temperatur magmatik. **[Optional Verification]** Sertakan saran untuk mengomparasikan hasil MT ini dengan data tomografi gelombang seismik ($V_p/V_s$) pada kedalaman yang sama.

## 4. Studi Kasus: Penginderaan Jauh Geofisika (InSAR) untuk Pemantauan Amblesan Tanah
Kamu bertindak sebagai **[Role]** Spesialis Geodesi Geofisika dan Satelit Radar. **[Task]** Lakukan analisis interferogram radar (InSAR) *time-series* untuk menghitung laju penurunan muka tanah (*land subsidence rate*) tahunan di kawasan pesisir. **[Context]** Kota pesisir mengalami banjir rob berulang akibat kombinasi kenaikan muka laut dan penurunan tanah akibat ekstraksi air tanah berlebih. Diberikan **[Input]** peta fase interferogram dinamis satelit Sentinel-1 yang menunjukkan pola *fringe* melingkar konsentris di kawasan industri dengan laju deformasi hingga -12 cm/tahun, gunakan **[Constraints]** untuk menghitung laju amblesan dalam satuan cm/tahun dan kaitkan dengan proses pemadatan (*kompaksi*) lapisan akuifer tertekan. Hasilkan **[Output]** berupa laporan teknis pemantauan deformasi permukaan beserta saran zonasi pembatasan pengambilan air tanah. **[Quality Criteria]** Analisis harus secara ilmiah mengaitkan perubahan fase jarak pandang satelit (*Line of Sight*) dengan perubahan volume fisik di bawah permukaan. **[Optional Verification]** Instruksikan verifikasi langsung di lapangan menggunakan stasiun pengukuran GNSS/GPS Kontinu (*CORS*).

## 5. Studi Kasus: Sub-Bottom Profiling (SBP) untuk Survei Kabel Bawah Laut
Kamu ditugaskan sebagai **[Role]** Geofisikawan Kelautan (*Marine Geophysicist*). **[Task]** Interpretasikan rekaman seismik pantul dangkal (*Sub-Bottom Profiling*/SBP) dan *Side Scan Sonar* untuk memetakan morfologi dasar laut dan potensi bahaya geologi (*geohazard*). **[Context]** Proyek penanaman kabel serat optik bawah laut antar-pulau membutuhkan rute yang aman dari paparan batuan keras, palung dangkal, dan aktivitas jangkar kapal. Berdasarkan **[Input]** rekaman SBP yang menunjukkan lapisan sedimen lumpur halus setebal 4 meter di atas batuan dasar (*basement*) keras serta citra *Side Scan Sonar* yang menampilkan bekas seretan jangkar (*anchor scars*), terapkan **[Constraints]** dengan membatasi analisis pada rekomendasi kedalaman penanaman kabel (*burial depth*) yang aman dari gangguan mekanis. Susun **[Output]** berupa laporan rekomendasi rute kabel (*Route Position List*) beserta peta rintangan dasar laut. **[Quality Criteria]** Penjelasan harus mengaitkan tingkat penetrasi gelombang akustik SBP dengan jenis material dasar laut serta identifikasi rintangan dari pantulan balik (*backscatter*). **[Optional Verification]** Sarankan verifikasi kondisi fisik sampel dasar laut menggunakan alat *Grab Sampler* atau visualisasi kamera ROV (*Remotely Operated Vehicle*).
